## 1. Calculadora de l'índex de massa corporal


Escriu una funció que calculi l'IMC ingressat per l'usuari/ària, és a dir, qui ho executi haurà d'ingressar aquestes dades. Pots obtenir més informació del seu càlcul en: 🔗 Índice de masa corporal IMC que es y como se calcula.

La funció ha de classificar el resultat en les seves respectives categories.

Consell: Intenta validar les dades prèviament, perquè enviï un missatge d’advertència si les dades introduïdes per l’usuari/a no estan en el format adequat o no pren valors raonables.

```python
import numpy as np
import pandas as pd
import string
import pyperclip
```

```python

def calculadora_IMC(pes, altura):
    """
    Calculadora del IMC. El pes ha de ser KG i l'altura en metres de la forma (1.78)
    """
    return round(pes / pow (altura,2), 2)


def classificar_imc(categoria):
    """
    Classificació IMC
    """

    if categoria < 18.5:
        return "Baix Pes"
    elif 18.5 <= categoria <= 24.9:
        return "Normal"
    elif 25 <= categoria <= 29.9:
        return "Sobrepes"
    else:
        return "Obesitat"
    

def obtenir_dades():
    try:
        pes = float(input("Introdueix el teu pes en kg: "))
        altura = float(input("Introdueix la teva alçada en metres: "))

    except ValueError:
        raise ValueError("Error: has d'introduir números vàlids.")

    if pes <= 0 or altura <= 0:
        raise ValueError("Error: has d'introduir números vàlids.")
    
    if altura > 2.5 or pes > 300:
        raise ValueError("Error: has d'introduir números vàlids.")

    return pes, altura


def execucio_imc():
    try:
        pes, altura = obtenir_dades()
        resultat_imc = calculadora_IMC (pes, altura)
        categoria = classificar_imc(resultat_imc)

        print(f"El teu Índex de Massa Corporal (IMC), amb un pes de {pes} KG i una altura de {altura} metres, és de : {resultat_imc} | El teu estat corporal és: {categoria}")
    
    except ValueError as e:
        print("Error:", e)



execucio_imc()
```

 ## 2. Convertidor de temperatures

 Existeixen diverses unitats de temperatura utilitzades en diferents contextos i regions. Les més comunes són Celsius (°C), Fahrenheit (°F) i Kelvin (K). També existeixen altres unitats com Rankine (°Ra) i Réaumur (°Re).

Selecciona almenys 2 conversors, de tal manera que en introduir una temperatura retorni, com a mínim, dues conversions, de manera que es puguin guardar (recorda que un print() no es pot guardar mai).

Consell: Intenta validar les dades prèviament, perquè enviï un missatge d’advertència si les dades introduides per l’usuari/a no estan en el format adequat.

(EXTRA): Pensa una manera d’emmagatzemar totes les possibles conversions en un sol objecte (Llista? Diccionari? DataFrame?) en comptes d’escriure molts if else en funció de la temperatura d’origen i la temperatura de destí.

```python

def convertir_temperatura(temperatura, origen):

    conversions = {}

    if origen == "C":

        conversions["Fahrenheit"] = round((temperatura * 9/5) + 32, 2)
        conversions["Kelvin"] = round(temperatura + 273.15, 2)

    elif origen == "F":

        celsius = (temperatura - 32) * 5/9

        conversions["Celsius"] = round(celsius, 2)
        conversions["Kelvin"] = round(celsius + 273.15, 2)


    elif origen == "K":

        celsius = temperatura - 273.15

        conversions["Celsius"] = round(celsius, 2)
        conversions["Fahrenheit"] = round((celsius * 9/5) + 32, 2)

    return conversions



def demanar_temperatura():

    temperatura = float(input("Introdueix la temperatura: "))

    return temperatura


def demanar_escala():

    escala = input("Introdueix l'escala (C si és Celsius, F si és Fahrenheit  o K si és Kelvin): ")

    escala = escala.upper()

    return escala


def validar_escala(escala):

    return escala in ["C", "F", "K"]



def executar_programa_temperatures():

    try:

        
        temperatura = demanar_temperatura()
        escala = demanar_escala()

       
        if not validar_escala(escala):

            print("Error: escala no vàlida.")

            return

        
        resultat = convertir_temperatura(temperatura,escala)


        print(f"Les equivalències de la temperatura {temperatura} d'escala {'Celsius' if escala == 'C' else 'Kelvin' if escala == 'K' else 'Fahrenheit'} són:")
        print(resultat)

    except:

        print("Error: has d'introduir una temperatura numèrica.")




executar_programa_temperatures()
```

## 3. Comptador de paraules d'un text.

Escriu una funció que donat un text, mostri les vegades que apareix cada paraula. Intenta que es gestionin totes les casuístiques possibles que facin que el programa no funcioni correctament.

(EXTRA): Quina és la llargada mitjana de les paraules del text que has escrit? “Hola com va?” hauria de retornar (4+3+2) / 3 = 3

```python

def netejar_text(text):

    text = text.lower()

    for signe in string.punctuation:

        text = text.replace(signe, "")

    return text


def obtenir_paraules(text):

    return text.split()



def comptar_paraules(paraules):

    comptador = {}

    for p in paraules:

        if p in comptador:

            comptador[p] += 1

        else:

            comptador[p] = 1

    return comptador



def calcular_mitjana(paraules):

    total_lletres = 0

    for p in paraules:

        total_lletres += len(p)

    if len(paraules) > 0:

        return round(total_lletres / len(paraules), 2)

    else:

        return 0



def processar_text(text):

    text_net = netejar_text(text)

    paraules = obtenir_paraules(text_net)

    frequencia = comptar_paraules(paraules)

    mitjana = calcular_mitjana(paraules)

    return frequencia, mitjana


def demanar_text():

    text = input("Escriu un text per comptar-ne les paraules: ")

    text = text.strip()

    return text



def executar_comptador_paraules():

    try:

       
        text_usuari = demanar_text()

        if text_usuari == "":

            print("Error: no has introduït cap text.")

            return

       
        frequencia, mitjana = processar_text(text_usuari)

        
        print("\nFreqüència de paraules:")
        print(frequencia)

        print(f"La llargada mitjana de les paraules del text és: {mitjana}") 


    except:

        print("S'ha produït un error.")



executar_comptador_paraules()
```

## 4. Diccionari invers (amb possibilitat de duplicats)

Resulta que el client té una enquesta molt antiga que s'emmagatzema en un diccionari i els resultats els necessita al revés, és a dir, intercanviant les claus i els valors. Els valors i claus en el diccionari original són únics; si aquest no és el cas, la funció hauria d'imprimir un missatge d'advertiment, juntament amb una llista amb els valors associats a la clau repetida.

```python

def invertir_diccionari(enquesta):
    invers = {}
    claus_repetides = {}
    hi_ha_duplicats = False

    for clau, valor in enquesta.items():
        if valor in invers:
            
            hi_ha_duplicats = True
        
            if not isinstance(invers[valor], list):
                invers[valor] = [invers[valor]]
            invers[valor].append(clau)  
            claus_repetides[valor] = invers[valor]
        else:
            invers[valor] = clau  
            
    if hi_ha_duplicats:
        print(f"\n ADVERTÈNCIA: S'han trobat valors repetits: {claus_repetides}")
    

    return invers


enquesta = {'a': 1, 'b': 2, 'c': 3}
enquesta_inversa = invertir_diccionari(enquesta)
print("\n Diccionari resultant:", enquesta_inversa)



enquesta = {'x': 'apple', 'y': 'banana', 'z': 'banana'}
enquesta_inversa = invertir_diccionari(enquesta)
print("\n Diccionari resultant:", enquesta_inversa)


```

## 1. Comptador i endreçador de paraules d'un text.

El client va quedar content amb el comptador de paraules, però ara vol llegir arxius TXT i que calculi la freqüència de cada paraula ordenades dins de les entrades habituals del diccionari segons la lletra amb la qual comencen, és a dir, les claus han d'anar de la A a la Z i dins de la A hem d'anar de la A la Z. Per exemple, per a l'arxiu "tu_me_quieres_blanca.txt" la sortida esperada seria:

```python
def llegir_fitxer(nom_arxiu):

    with open(nom_arxiu, "r", encoding="utf-8") as f:

        text = f.read()

    return text


def netejar_text(text):

    text = text.lower()

    for signe in string.punctuation:

        text = text.replace(signe, "")

    return text



def obtenir_paraules(text):

    return text.split()



def comptar_ordenar_paraules(paraules):

    diccionari = {}

    
    for paraula in sorted(paraules):

        
        lletra_inicial = paraula[0]

        if lletra_inicial not in diccionari:

            diccionari[lletra_inicial] = {}

        if paraula in diccionari[lletra_inicial]:

            diccionari[lletra_inicial][paraula] += 1

        else:

            diccionari[lletra_inicial][paraula] = 1

    return diccionari


def processar_fitxer(nom_arxiu):

   
    text = llegir_fitxer(nom_arxiu)

    text_net = netejar_text(text)

    paraules = obtenir_paraules(text_net)

    resultat = comptar_ordenar_paraules(paraules)

    return resultat


processar_fitxer("tu me quieres blanca.txt")
```

## 2. Conversió de tipus de dades.
El client rep una llista de dades i necessita generar dues llistes, la primera on estaran tots els elements que es van poder convertir en flotants i l'altra on estan els elements que no es van poder convertir. Exemple de la llista que rep el client:


```python

def convertir_a_float (llista):

    convertibles = []
    no_convertibles = []

    for element in llista:

        if isinstance(element, (list, tuple, set)):

            # Recorrem els elements interns
            for item in element:

                try:

                    numero = float(item)
                    convertibles.append(numero)

                except:

                    no_convertibles.append(item)

        else:

            try:

                numero = float(element)
                convertibles.append(numero)

            except:

                no_convertibles.append(element)

    return convertibles, no_convertibles



llista_entrada = ['1.3', 'one', '1e10', 'seven', '3-1/2', ('2', 1, 1.4, 'not-a-number'), [1, 2, '3', '3.4']]
 
convertits, no_convertits = convertir_a_float(llista_entrada)
 
print(f"\n Llista original: {llista_entrada}")
print(f"\n Elements convertits a float: {convertits}")
print(f"\n Elements que no s'han pogut convertir a float: {no_convertits}")

```

## 1. Generador de Contrasenyes

Explora el funcionament del mòdul random de la llibreria numpy. 🔗 Random module in NumPy

En aquest punt, el client ha detectat un problema amb les contrasenyes que utilitzen els seus treballadors. Asdf1234, dates d’aniversari o similars. Per a solucionar-ho ens ha encarregat una funció de Python que generi contrasenyes més segures. La funció ha de dependre dels següents paràmetres:

longitud (int): Longitud de la contrasenya
majuscules (bool = True): Si hi ha d’aparèixer majúscules
minuscules (bool = True): Si hi ha d’aparèixer minúscules
numeros (bool = True): Si hi ha d’aparèixer números
signes (bool = False): Si hi ha d’aparèixer caràcters especials (,-$? o similars)
Així doncs, si executem la funció de la següent manera:

crear_contrasenya(10, True, True, True, True)

Hauríem d’obtenir un output (que hem de poder guardar) de l’estil:

9Er,5Vn8P$

Assegura’t que es compleixin tots els criteris, i que aquestes contrasenyes siguin realment aleatòries.

(EXTRA) Explora com podríem fer que la funció copiés la contrasenya automàticament al porta papers de l’ordinador (com si l’haguéssim seleccionat i fet ctrl+copy).

```python
def definir_contraseña_mínima (majuscules, minuscules, numeros, signes):
    """Crea la cadena amb tots els caràcters permesos i inicialitzem una llista amb els tipus de caràcter que han d'aparèixer mínim un cop en la contrasenya"""
    tots_caracters = ""
    obligatoris = []
    
    if majuscules:
        tots_caracters += string.ascii_uppercase
        obligatoris.append(np.random.choice(list(string.ascii_uppercase)))
    if minuscules:
        tots_caracters+= string.ascii_lowercase
        obligatoris.append(np.random.choice(list(string.ascii_lowercase)))
    if numeros:
        tots_caracters += string.digits
        obligatoris.append(np.random.choice(list(string.digits)))
    if signes:
        tots_caracters += string.punctuation
        obligatoris.append(np.random.choice(list(string.punctuation)))
        
    return tots_caracters, obligatoris

def crear_contrasenya(longitud, majuscules=True, minuscules=True, numeros=True, signes=False):
    """Funció principal que orquestra la creació de la contrasenya."""

    tots_caracters, contrasenya = definir_contraseña_mínima (majuscules, minuscules, numeros, signes)

  
    if len(tots_caracters) == 0:
        return "Error: has de seleccionar almenys un tipus de caràcter."
    if longitud < len(contrasenya):
        return f"Error: la longitud mínima ha de ser de {len(contrasenya)} per complir els criteris."

    llista_caracters = list(tots_caracters)
    while len(contrasenya) < longitud:
        contrasenya.append(np.random.choice(llista_caracters))

  
    np.random.shuffle(contrasenya)
    contrasenya_final = "".join(contrasenya)

    pyperclip.copy(contrasenya_final)
    
    return contrasenya_final


nova_contrasenya = crear_contrasenya(10, True, True, True, True)

print(f"Contrasenya generada (i copiada al portapapers): {nova_contrasenya}") 
```

## 2. Processament de dades simple

Una companya de feina ens ha demanat un favor, aprofitant que sap que estem aprenent a programar. Té un històric de partits de futbol català en un fitxer, on hi ha emmagatzemat els noms dels equips i els resultats. Necessita que processem les dades de manera automàtica, per extreure’n els resultats que necessita.

Utilitza el fitxer "historic_partits.txt"

Necessita un programa que retorni:

El nombre total de gols que ha fet cada equip.
El nom de l’equip més golejador.
El nom de l’equip més golejat
La classificació global (cada victòria: 3 pts, empat 1 pts, derrota 0 pts)
Com a referència, els equips amb més punts serien els següents:

```python

def llegir_fitxer(nom_arxiu):
    """
    Obre el fitxer i retorna una llista de línies no buides.
    """
    with open(nom_arxiu, "r", encoding="utf-8") as f:
        linies = f.readlines()
   
    linies_netes = [linia.strip() for linia in linies if linia.strip()]
    return linies_netes
 

def tractament_linia(linia):
    """
    Funció que retorna les dades: equip_local, equip_visitant, gols_local, gols_visitant
    
    """
    parts = linia.split()
 
    index_resultat = None
    for i, part in enumerate(parts):
        if "-" in part and part.replace("-", "").isdigit():
            index_resultat = i
            break
 
    if index_resultat is None:
        print(f"No he pogut llegir la línia: '{linia}'")
        return None
 
    equip_local    = " ".join(parts[:index_resultat])
    equip_visitant = " ".join(parts[index_resultat + 1:])
   
    gols = parts[index_resultat].split("-")

    gols_local = int(gols[0])

    gols_visitant = int(gols[1])
 
    return equip_local, equip_visitant, gols_local, gols_visitant
 
 

def actualitzar_dades(dades, equip, gols_marcats, gols_rebuts, resultat):
    """
    Actualitza el diccionari 'dades' per a un equip concret.
 
    Paràmetres:
    - dades       : diccionari acumulador { equip: [G, E, P, gols, enc, pts] }
    - equip       : nom de l'equip a actualitzar
    - gols_marcats: gols que ha marcat aquest equip en el partit
    - gols_rebuts : gols que ha rebut aquest equip en el partit
    - resultat    : 'V' (victòria), 'E' (empat) o 'D' (derrota)
    """

    if equip not in dades:
        dades[equip] = [0, 0, 0, 0, 0, 0]
    
 
    dades[equip][3] += gols_marcats
    dades[equip][4] += gols_rebuts
 
    if resultat == "V":
        dades[equip][0] += 1   
        dades[equip][5] += 3   
    elif resultat == "E":
        dades[equip][1] += 1   
        dades[equip][5] += 1   
    elif resultat == "D":
        dades[equip][2] += 1 
 
 
def processar_partits(linies):
    """
    Rep la llista de línies i retorna el diccionari amb totes les dades acumulades dels equips 
    """
    dades = {}
 
    for linia in linies:
        resultat_tractat = tractament_linia(linia)
 
        if resultat_tractat is None:
            continue  # Línia invàlida, la saltem
 
        equip_local, equip_visitant, gols_local, gols_visitant = resultat_tractat
 
      
        if gols_local > gols_visitant:
            res_local, res_visitant = "V", "D"
        elif gols_local < gols_visitant:
            res_local, res_visitant = "D", "V"
        else:
            res_local, res_visitant = "E", "E"
 
        actualitzar_dades(dades, equip_local, gols_local, gols_visitant, res_local)
        actualitzar_dades(dades, equip_visitant, gols_visitant, gols_local, res_visitant)
    
 
    return dades
 
 
def construir_dataframe(dades):
    """
    Converteix el diccionari acumulador en un DataFrame ordenat per punts.
    """
    df = pd.DataFrame.from_dict(
        dades,
        orient="index",
        columns=["partits guanyats", "partits empatats", "partits perduts",
                 "gols", "gols encaixats", "punts"]
    )
    df = df.sort_values("punts", ascending=False)
    return df
 
 

def extreure_estadistiques(df):
    """
    A partir del DataFrame, retorna:
    - gols_dict    : diccionari { equip: gols_marcats }
    - mes_golejador: nom de l'equip que ha marcat més gols
    - mes_golejat  : nom de l'equip que ha rebut més gols
    """
    gols_dict     = df["gols"].to_dict()
    mes_golejador = df["gols"].idxmax()
    mes_golejat   = df["gols encaixats"].idxmax()
 
    return gols_dict, mes_golejador, mes_golejat
 
 

def mostrar_resultats(df, gols_dict, mes_golejador, mes_golejat):
    """
    Imprimeix per pantalla tots els resultats de forma clara.
    """

    print(" DATAFRAME D'ESTADÍSTIQUES:")
   
    print(df.to_string())
 
    print("\n NÚMERO DE GOLS PER EQUIP:")
    print(gols_dict)
 
    print(f"\n Equip més golejador : {mes_golejador}")
    print(f"\n Equip més golejat   : {mes_golejat}")



def main(nom_arxiu):
    linies        = llegir_fitxer(nom_arxiu)  
    dades         = processar_partits(linies)        
    df            = construir_dataframe(dades)         
    gols_dict, mes_golejador, mes_golejat = extreure_estadistiques(df)  
    mostrar_resultats(df, gols_dict, mes_golejador, mes_golejat)        
 
    return df, gols_dict, mes_golejador, mes_golejat
 

df, gols_dict, mes_golejador, mes_golejat = main("historic partits.txt")
 
```
