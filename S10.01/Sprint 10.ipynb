## 4.

o	Crea una taula resum que permeti veure el sou mig, medià, mínim i màxim per Gènere.

o	Ordena la taula en funció del sou mig.



```python

def netejar_salari(salari_text):
    """Funció per a transformar el salari mensual d'string a nombre"""
    if pd.isnull(salari_text) or isinstance(salari_text, (int, float)):
        return salari_text

    text = str(salari_text).strip()

    text = text.replace("€", "")
    text = text.replace(".", "")
    text = text.replace(",", ".").strip()

    try:
        return float(text)
    except ValueError:
        return np.nan



df_treballadors["Salari mensual"] = df_treballadors["Salari mensual"].apply (netejar_salari)

taula_salaris = (df_treballadors.groupby("Gènere")["Salari mensual"].agg(
          Sou_mig="mean",
          Sou_medià="median",
          Sou_mínim="min",
          Sou_màxim="max"
      ).sort_values(by="Sou_mig", ascending=False)
      ).round(2)
      
print("\nTaula de salaris  per Gènere (amb salaris convertits a float)")

taula_salaris
```

o	Utilitza la funció que acabes de crear per generar una columna nova al DataFrame amb l'edat actual.


```python
df_treballadors["Data_naixement"] = pd.to_datetime({
    "year": df_treballadors["Any"],
    "month": df_treballadors["Mes"],
    "day": df_treballadors["Dia"]
})
```

```python
def calcular_edat(data_naixament):
    data_avui = datetime.now()
    edat = data_avui.year - data_naixament.year
    if (data_avui.month, data_avui.day) < (data_naixament.month, data_naixament.day):
        edat -= 1

    return edat



df_treballadors["Edat_actual"] = df_treballadors["Data_naixement"].apply(calcular_edat)

df_treballadors.head()
```

Crea una funció que prengui un dataframe com a paràmetre d'entrada.
La funció ha de crear (i exportar) un gràfic automàticament per a cada columna del dataframe. Per exemple:
________________________________________
•	un histograma/boxplot si la variable és numèrica
•	unes barres dels valors més freqüents si és categòrica
•	unes barres dels anys més freqüents si la dada està en format data.

La idea és crear una funció que funcioni per qualsevol dataframe, no només amb el que hem treballat fins ara.
Mostra el resultat de la funció en algun dels datasets d’exemple que conté el paquet seaborn. Per exemple, iris, penguins o titanic.
Tingues en consideració que en el següent sprint treballaràs exclusivament amb gràfics. L’objectiu d’aquest exercici no és crear gràfics molt elaborats, sinó resoldre una necessitat de manera ràpida i automàtica.


```python
def crear_grafics(df):
    """
    Genera i desa automàticament gràfics segons el tipus de dada de cada columna.

    Parameters:df (pd.DataFrame): El DataFrame de Pandas que es vol analitzar.

    Comportament:
    - Numèriques: Crea una figura doble amb histograma i boxplot.
    - Dates: Crea un gràfic de barres amb la freqüència per anys.
    - Categòriques: Crea un gràfic de barres amb el Top 15 de valors.
    Cada gràfic es desa com a imatge '.png' amb el nom de la columna.
    """
    

    for columna in df.columns:
        
        
        plt.figure(figsize=(7, 4)) 
        
        if pd.api.types.is_numeric_dtype(df[columna]) and not pd.api.types.is_bool_dtype(df[columna]):  ##proporciona conjunt de funcions per avaluar el tipus de dada
            
            plt.subplot(1, 2, 1)
            

            df[columna].dropna().plot(kind='hist', bins=15) ##usant dropna per eliminar valors nuls
            plt.title(f"Histograma de {columna}")


            plt.subplot(1, 2, 2)
            
            plt.boxplot(df[columna].dropna())

            plt.title(f"Boxplot de {columna}")

    
        elif pd.api.types.is_datetime64_any_dtype(df[columna]):
            df[columna].dt.year.value_counts().sort_index().plot(kind="bar") # amb .dt.year extraiem l'any, amb .value_counts() contem les vegades que es repeteix l'any, ordenem l'any sort_index() 
            plt.title(f"Anys més freqüents {columna}")
        
      
        else:
            df[columna].value_counts().head(15).plot(kind="bar")
            plt.title(f"Valors més freqüents {columna}")
        
     
        plt.tight_layout()
        plt.savefig(f"grafic_{columna}.png")
        plt.show()
        plt.close()
        

df_penguins = sns.load_dataset("penguins")
crear_grafics(df_penguins)


```

### 2.

Carrega l'arxiu matriu_distancies.xlsx a pandas, de manera que els noms de files i els noms de columnes siguin els de les ciutats. Borra "Las Palmas de Gran Canaria" i "Palma" perquè poguem fer el trajecte en cotxe.
Font: Mejores Rutas

Ens interessa visitar totes les ciutats principals d'Espanya recorrent la mínima distància possible.
No cal que ho facis de forma òptima, ens interessa que desenvolupis una solució raonable utilitzant les eines que tens actualment.
Per exemple, una aproximació senzilla (que no òptima) seria anant sempre a la ciutat més propera que no haguem visitat encara
Fes una funció que donada la matriu de distàncies i la ciutat d'origen, faci una proposta de ruta que sigui el més curta possible que puguis, retornant una llista amb l'ordre de visita. Dóna també la distància total recorreguda.
(EXTRA) Des de quina ciutat la ruta seria més curta amb l'algoritme plantejat



```python
def carregar_matriu(nom_fitxer):
    """
    Llegeix l'arxiu Excel amb una matriu de distàncies i eliminem les ciutats "Las Palmas de Gran Canaria","Palma" del Datafrma
    Retornem el Dataframe eliminant les ciutats "Las Palmas de Gran Canaria","Palma" 
    """
    df = pd.read_excel(nom_fitxer, index_col=0)
    df = df.drop(['Palma', 'Las Palmas de Gran Canaria'], axis=0)
    df = df.drop(['Palma', 'Las Palmas de Gran Canaria'], axis=1)

    return df
   
def ciutat_mes_propera(df, ciutat_actual, visitades):

    """
    Busca la ciutat no visitada més propera des de la ciutat actual.
    Paràmetres:df (DataFrame): Matriu de distàncies, ciutat_actual (str):Ciutat on ens trobem actualment.visitades (list): lista de ciutats ja visitades.
    Retorna: ciutat més propera, distància fins aquesta ciutat
    """

    
    distancies = df.loc[ciutat_actual]
  
    millor_ciutat = None
    millor_distancia = 10000

    for ciutat in distancies.index:

        distancia = distancies[ciutat]

        if ciutat != ciutat_actual and ciutat not in visitades:

            if distancia < millor_distancia:

                millor_distancia = distancia
                millor_ciutat = ciutat

    return millor_ciutat, millor_distancia


def crear_ruta(df, ciutat_inicial):
        
    """
    Crea una ruta aproximada visitant sempre la ciutat no visitada més propera.
    Paràmetres: df (DataFrame):Matriu de distàncies, ciutat_inicial (str):Ciutat des d'on comença la ruta.
    Retorna: llista amb l'ordre de visita, distància total recorreguda
    """

    visitades = []
    ruta = []

    distancia_total = 0

    ciutat_actual = ciutat_inicial

    visitades.append(ciutat_actual)
    ruta.append(ciutat_actual)

    while len(visitades) < len(df.index):

 
        seguent_ciutat, distancia = ciutat_mes_propera(df,ciutat_actual,visitades)

        visitades.append(seguent_ciutat)
        ruta.append(seguent_ciutat)

        distancia_total += distancia

        ciutat_actual = seguent_ciutat

    return ruta, distancia_total



df = carregar_matriu("matriu_distancies.xlsx")
ciutat_inicial = "Barcelona"

ruta, distancia_total = crear_ruta(df, ciutat_inicial)

print(f"\nLa ruta proposada és: {ruta}") 
print(f"\nLa ruta te una distància de: {distancia_total} Km") 
```
