### 1. Consulta l’API pública JSONPlaceholder utilitzant el mètode GET per obtenir:

o	Llista de publicacions (/posts)

o	Llista d’usuaris (/users)

o	Llista de tasques (/todos)


### 2. Mostra per pantalla:

o	La quantitat total de cada recurs.

o	El codi d'estat de cada petició.





```python
url = "https://jsonplaceholder.typicode.com/"

endpoints = ["/posts", "/users", "/todos"]

for endpoint in endpoints:
    try:
  
        response = requests.get(f"{url}{endpoint}") 
        response.raise_for_status()
        
    
        data = response.json()
        print(f"\n{endpoint}\n{data}")
        print(f"Recurs {endpoint}: S'han obtingut {len(data)} elements. Codi: {response.status_code}")
        
    except requests.exceptions.HTTPError as err:
        print(f"Error HTTP en obtenir {endpoint}: {err}")
    except Exception as e:
        print(f"S'ha produït un error inesperat: {e}")
```

### 3. Fes una petició a una publicació inexistent per obtenir un error 404 i mostra el codi d'estat rebut.


```python
error_req = requests.get(f"{url}/error")
try:
    error_req.raise_for_status()
except requests.exceptions.HTTPError as e:
        print(f"Error HTTP: {e}")
```

### 4. Fes una petició POST per crear una nova publicació fictícia. Inclou el títol, el cos del missatge i un userId.

o	Mostra la resposta JSON.
o	Mostra el codi d'estat.


```python
nova_publicacio = {
    "title": "Primer post de prova",
    "body": "Aquest és un missatge creat amb Python",
    "userId": 122
}

resposta = requests.post("https://jsonplaceholder.typicode.com/posts", json=nova_publicacio)

print("JSON respuesta:", resposta.json())
print("Status:", resposta.status_code)
```

### 5. Fes una petició PATCH per modificar parcialment una publicació existent.
o	Mostra la resposta JSON.
o	Mostra el codi d'estat.


```python
canvi = {"title": "Títol modificat", "body": "Modificació feta per a l'exercici 5 del nivell 1" }

resposta = requests.patch(
    "https://jsonplaceholder.typicode.com/posts/101",
    json=canvi
)

print("JSON respuesta:", resposta.json())
print("Status:", resposta.status_code)
```

### 6. Fes una petició DELETE sobre una publicació.
o	Mostra la resposta JSON.
o	Mostra el codi d'estat.


```python
response_delete = requests.delete(f"{url}/posts/101")

print(response_delete.json())
print("Status:", response_delete.status_code)
```

3. date

**Verifica que la resposta sigui en format JSON (és el més adequat per a aquest exercici).**

```python
response = requests.get("https://api.tvmaze.com/search/shows")
print(response.headers["Content-Type"])
```

Mirem la capçalera de "Content-Type" del protocol HTTP per veure el format en el que s'envien les dades. Com que veiem que retorna "application/json; charset=UTF-8" indica que la resposta esà en format JSON

### 3. Fes una petició GET senzilla:

o	Mostra el codi d'estat de la resposta.

o	Imprimeix de forma clara alguns dels camps de la resposta JSON.


```python
# Cercarem sèries que continguin "Batman"
url_2 = "https://api.tvmaze.com/search/shows"
params = {'q': 'batman'}

try:
    response = requests.get(url_2, params=params)
    response.raise_for_status() 
    print(f"Codi d'estat: {response.status_code}")
    
    data = response.json()
    

    for entry in data[:3]:
        show = entry['show']
        print(f"Sèrie: {show['name']}, Gèneres: {', '.join(show['genres'])}")
        
except requests.exceptions.HTTPError as err:
    print(f"Error HTTP: {err}")
```

### 4. Converteix les dades a un DataFrame de pandas:

o	Mostra les primeres files del DataFrame.


```python
# Extraiem només la informació de 'show' per a cada resultat
show_list = [entry['show'] for entry in data]
df = pd.DataFrame(show_list)


df.head()
```

## Nivell 3

[API d'Open Data Barcelona](https://opendata-ajuntament.barcelona.cat/ca/desenvolupadors#APIS)

### 1. Utilitza l’API d’Open Data BCN per:

1.a.	Cercar un dataset del teu interès mitjançant package_search


```python
url_3 = "https://opendata-ajuntament.barcelona.cat/data/api/action/"
params_search = {"q":"multes"}

resposta_search = requests.get(f"{url_3}/package_search", params=params_search) 

try:
    resposta_search.raise_for_status()
    print(f"Status: {resposta_search.status_code}")

    resultats = resposta_search.json()

    print("Datasets trobats:")
    print(len(resultats["result"]["results"]))

    datasets = resultats["result"]["results"]

    for ds in datasets[:10]:
        print("Títol:", ds["title"])
        print("ID:", ds["id"])
        print("--------------")

except requests.exceptions.HTTPError as e:
        print(f"Error HTTP: {e}")

```

*1.B package_show*

```python

##Agafem el id del quart dataset dels obtinguts per exemple
dataset_id = datasets[0]["name"]

print (dataset_id)


url_4 = "https://opendata-ajuntament.barcelona.cat/data/api/action/package_show"

params = {"id": dataset_id}

resposta_show = requests.get(url_4, params=params)
resposta_show.raise_for_status()

dades_show = resposta_show.json()


print("Títol:", dades_show["result"]["title"])
print("Número de recursos:", len(dades_show["result"]["resources"]))

# Mostrar los primeros 3 recursos
for recurs in dades_show["result"]["resources"]:
    print("Nombre:", recurs["name"])
    print("Formato:", recurs["format"])
    print("Resource ID:", recurs["id"])
    print("-------------")

```

### 2. Dels resultats obtinguts, selecciona’n un que tingui recursos disponibles en CSV o JSON.


```python
resource_id = None
recursos = dades_show["result"]["resources"]

for recurs in recursos:
    format_recurs = recurs["format"].lower()
    
    if format_recurs in ["csv", "json"]:
        resource_id = recurs["id"]
        print(f"\nRecursos seleccionat: {recurs['name']} (Format: {recurs['format']})")
        print(f"Resource ID: {resource_id}")
        break
```

### 3. Amb el resource_id del recurs seleccionat, realitza una consulta mitjançant datastore_search per:

o	Recuperar almenys 100 registres.


```python
if resource_id:
    url_datastore = "https://opendata-ajuntament.barcelona.cat/data/api/action/datastore_search"
    
    params_datastore = {
        "resource_id": resource_id,
        "limit": 100
    }
    
    resposta_ds = requests.get(url_datastore, params=params_datastore)
    resposta_ds.raise_for_status()
    
    dades_records = resposta_ds.json()
    
    llista_registres = dades_records["result"]["records"]
    
    print(f"\nS'han recuperat {len(llista_registres)} registres correctament.")
    
    if llista_registres:
        print("Exemple del primer registre:", llista_registres[0])
else:
    print("No s'ha trobat cap recurs en format CSV o JSON en aquest dataset.")
```

### 4. Converteix els resultats a un DataFrame.


```python
df = pd.DataFrame(llista_registres)
df
```

```python
df.to_csv("tipus_vehicles_multes.csv", index=False)
```
